```python
import requests
import random
from colorama import Fore, Back, Style

print("You have six attempts to guess a random five-letter word.")
print("Every guessed letter in it's position marken green, guessed letter not in it's position — yellow, otherwise — grey.")


def get_wordle_words() -> str:
  response = requests.get('https://api.frontendexpert.io/api/fe/wordle-words')
  wordle_words = response.json()
  return random.choice(wordle_words)


def game():
  word = get_wordle_words()

  attempts = 0
  mistakes = 0

  while attempts < 6:
    if mistakes != 3:
      users_word = str(input("Enter a five-letter word: "))
      users_word = users_word.upper()

      if len(users_word) != 5:
        mistakes += 1
        print("Strictly five-letter words. Your attempt won't be encounted.")
        if mistakes != 3:
          print(f"{3 - mistakes} more mistake(s) and you this game will be ceased")
          print("")
      else:
        output = ""

        for pos, letter in enumerate(users_word):
          if not letter.isalpha():
            print("There is at least one incorrect symbol")
            attempts -= 1

          if letter == word[pos]:
            output += Back.GREEN
          elif letter in word:
            output += Back.YELLOW
          else:
            output += Back.LIGHTBLACK_EX

          output += Fore.WHITE + " " + letter + " " + Style.RESET_ALL

        print(output)
        print("")

        attempts += 1
        if users_word == word:
          print("Congratulations! You won!")
          attempts = 7
    else:
      print("")
      print(Fore.WHITE + Back.RED + "Too much mistakes, you sabotaged this game" + Style.RESET_ALL)
      if attempts > 0:
        print(f"The word was {word}")
      attempts = 7

  if attempts == 6:
    print(Fore.WHITE + Back.RED + "Attempts are over!" + Style.RESET_ALL)
    print(f"The word was {word}")



print("")
is_ready = str(input("Are you ready? (YES or NO): "))
is_ready = is_ready.upper()

if is_ready == "YES":
  print("Good luck!")
  print("")
  game()


while is_ready == "YES":
  print("")
  is_ready = str(input("One more time? (YES or NO): "))
  is_ready = is_ready.upper()

  if is_ready == "YES":
    print("Good luck!")
    print("")
    game()
  else:
    print("")
    print("Thanks for game")
    break


print("See you next time!")
```
